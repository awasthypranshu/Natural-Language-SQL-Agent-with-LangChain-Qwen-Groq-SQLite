```python
import streamlit as st
from pathlib import Path
from langchain_community.agent_toolkits import create_sql_agent
from langchain_community.utilities import SQLDatabase
from sqlalchemy import create_engine
import sqlite3
from langchain_groq import ChatGroq
from langchain_community.agent_toolkits import SQLDatabaseToolkit
```

```python
llm=ChatGroq(model="qwen/qwen3.6-27b",max_tokens=800)
```

```python
def configure_sql():
    file_path = Path("student.db").absolute()
    creator =lambda: sqlite3.connect(f"file:{file_path}?mode=ro", uri=True)
    engine = create_engine("sqlite:///",creator = creator)
    return SQLDatabase(engine)
```

```python
toolkit = SQLDatabaseToolkit(llm=llm, db=configure_sql())
```

```python
agent = create_sql_agent(
    llm=llm,
    toolkit=toolkit,
)
res = agent.invoke("Which student has the highest marks?")
res
```
